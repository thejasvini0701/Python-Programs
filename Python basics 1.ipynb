```python
#Calculate the compound interest for a given principal amount, interest rate, and time period
principal=float(input("Enter principal amouunt:"))
interest=float(input("Enter interest rate:"))
time=float(input("Enter time period:"))
amount=principal*(1+interest/100)**time
compound_interest=amount-principal
print("Compound interest:",compound_interest)
```
