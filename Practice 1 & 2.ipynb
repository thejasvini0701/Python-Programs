```python
#Take an employee's basic salary.
#HRA = 20% of basic salary
#DA = 10% of basic salary
#Calculate the gross salary.
basic_salary=30000
HRA=(basic_salary*20)/100
DA=(basic_salary*10)/100
print("HRA:",HRA)
print("DA:",DA)
print("Gross salary:",basic_salary+HRA+DA)
```

```python
#Take the prices of three products and calculate the total bill.If the total is above ₹2000, give a 10% discount.
p1=800
p2=700
p3=900
total=p1+p2+p3
if total>2000:
    discount=(total*10)/100
else:
    discount=0
final_bill=total-discount
print("Total:",total)
print("Discount:",discount)
print("Final_bill:",final_bill)
```

```python
#Calculate delivery charges based on order amount:
#Order ≥ ₹2000 → Free delivery
#₹1000–₹1999 → ₹50 delivery
#Below ₹1000 → ₹100 delivery
order_amount=1350
if order_amount>=2000:
    delivery=0
elif order_amount>=1000:
    delivery=50
else:
    delivery=100

final_amount=order_amount+delivery
print('delivery_charge:',delivery)
print("final_amount:",final_amount)
```
