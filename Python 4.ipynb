```python
#Write a Python function to check if a given string is a palindrome
def palindrome(s):
    rev=s[::-1]
    return rev
word="madam"
if palindrome(word):
    print("palindrome")
else:
    print("Not a palindrome")
print("Palindrome of given string:",palindrome(word))
```

```python
# Implement a function that returns the factorial of a given number using recursion
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fact(n-1)
num=5
print(fact(num))
        
```

```python
#Create a function to find the square of each element in a given list
l1=[1,2,3]
def square(lst):
    l2=[]
    for i in lst:
        sq=i**2
        l2.append(sq)
    return l2
print("squares of each element in list:",square(l1))
```

```python
#Write a function that takes two lists and returns their intersection (common elements)
a=[2,3,2,4,5]
b=[4,3,1,7,6]
def inter(a,b):
    c=[]
    for i in a:
        if i in b and i not in c:
            c.append(i)
    return c
print("common elements:",inter(a,b))
```

```python
#Implement a function to check if a given year is a leap year or not
yr=2026
def leap(yr):
    if yr%400==0 and yr%4==0:
        return True
    elif yr%100==0:
        return False
    else:
        return False
if leap(yr):
    print("It is leap year")
else:
    print("It is not leap year")
        
```
