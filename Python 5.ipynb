```python
#Write a function to count the number of vowels in a given string
def vowels(str):
    vowels='aeiouAEIOU'
    count=0
    for i in str:
        if i in vowels:
            count+=1
    return count
str="Owl"
print("Number of vowels in string:",vowels(str))
```

```python
# Implement a function that checks if a given string is a pangram (contains all letters of the alphabet)
def pangram(s):
    alphabet='abcdefghijklmnopqrstuvwyz'
    for i in alphabet:
        if i not in s.lower():
            return False
    return True
s='Hello'
if pangram(s):
    print('It is panagram')
else:
    print("It is not pangram")
```

```python
#Given a string, write a function to remove all vowels from it and return the modified string
def remove_vowel(str):
    vowels='aeiouAEIOU'
    result=""
    for i in str:
        if i in vowels:
            result+=i
    return result
print("String without vowels:",remove_vowel(str))
```

```python
# Create a function that takes a sentence as input and returns the sentence in reverse order
def reverse_string(sentence):
    words=sentence.split()
    return " ".join(words[::-1])
print("Reversed string:",reverse_string(sentence))
```

```python
#Write a function to remove all duplicate characters from a given string
def remove_duplicates(text):
    result=""
    for i in text:
        if i not in result:
            result+=i
    return result
text="hello hey hello"
print("String without duplicates:",remove_duplicates(text))
```
