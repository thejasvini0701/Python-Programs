```python
#Implement a program that finds the largest numberin a list
l=[10,3,2,5,45]
largest=l[0]
for i in l:
    if i>largest:
        largest=i
        
print("Largest number:",largest)
```

```python
# Create a program that generates the Fibonacci sequence up to a given number of terms
n=int(input("Enter number of terms:"))
a=0
b=1
for i in range(n):
    print(a)
    a,b=b,a+b
```

```python
#Write a program that calculates the factorial of a given number
a=5
def fact(a):
    if a==0 or a==1:
        return 1
    else:
        return a*fact(a-1)
print(fact(a))
        
```

```python
# Create a loop that prints all prime numbers between 1 and 50
for i in range(2,51):
    is_prime=True
    for n in range(2,i):
        if i%n==0:
            is_prime=False
            break
    if is_prime:
        print(i)
    
```

```python
#Given a list of words, count the number of words with more than five characters
words=["apple", "banana", "cat", "elephant", "school", "computer"]
count=0
for i in words:
    if len(i)>5:
        count=count+1
print("Numbers of words:",count)
        
```
